```python
import pandas as pd
import pandas.io.sql as pds
import sqlite3 as sq3
```

```python
import os
import urllib.request

# Initialize path to SQLite
path = "classic_rock.db"
data = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-ML0232EN-SkillsNetwork/asset/classic_rock.db"

if not os.path.exists(path) or os.path.getsize(path) == 0:
    urllib.request.urlretrieve(data, path)

con = sq3.Connection(path)
```

```python
query = "SELECT * FROM rock_songs"
df = pd.read_sql(query, con)
```

```python
# Create a query to count the number of songs by each artist and release year
query = '''
SELECT Artist, Release_Year, COUNT(*) AS num_songs, AVG(PlayCount) AS avg_plays  
    FROM rock_songs
    GROUP BY Artist, Release_Year
    ORDER BY num_songs desc;
'''

# Execute the query and store the results in a DataFrame

observations = pds.read_sql(query, con)

observations.head()
```
